```python
import socket

client = socket.socket(socket.AF_INET, socket.SOCK_DGRAM)

message = "Hello from Cyber Security Lab!"

client.sendto(message.encode(), ("127.0.0.1", 9999))

data, address = client.recvfrom(1024)

print("Server:", data.decode())

client.close()
```
