```python
# Program 6: Banner Grabbing
## Statement
Write a Python program to connect to a service running on localhost
and retrieve its initial banner information.
```

```python
import socket

HOST = "127.0.0.1"
PORT = 9999

client = socket.socket(socket.AF_INET, socket.SOCK_STREAM)
client.settimeout(3)

try:
    client.connect((HOST, PORT))

    banner = client.recv(1024).decode(errors="ignore")

    print("===== BANNER INFORMATION =====")
    print("Host:", HOST)
    print("Port:", PORT)
    print("Banner:", banner)

except ConnectionRefusedError:
    print("No service is running on the selected port.")

except socket.timeout:
    print("Connection timed out.")

finally:
    client.close()
```

```python
# Program 10: Simple HTTP Server
## Statement
Write a Python program to create a simple local HTTP server
and serve a custom cybersecurity webpage.
```

```python
from http.server import HTTPServer, BaseHTTPRequestHandler


class CyberHandler(BaseHTTPRequestHandler):

    def do_GET(self):

        message = """
        <!DOCTYPE html>
        <html>
        <head>
            <title>Cyber Security Lab</title>
        </head>
        <body>
            <h1>Welcome to the Cyber Security Lab</h1>
            <p>Local HTTP Server is running successfully.</p>
            <p>Status: Secure Lab Environment</p>
        </body>
        </html>
        """

        try:
            self.send_response(200)
            self.send_header("Content-Type", "text/html")
            self.send_header("Content-Length", str(len(message.encode())))
            self.end_headers()

            self.wfile.write(message.encode())

        except ConnectionAbortedError:
            print("Client closed the connection.")

        except ConnectionResetError:
            print("Client reset the connection.")


server = HTTPServer(("127.0.0.1", 8080), CyberHandler)

print("===================================")
print("   CYBER SECURITY HTTP SERVER")
print("===================================")
print("Server running at:")
print("http://127.0.0.1:8080")
print("Press CTRL+C to stop the server.")

try:
    server.serve_forever()

except KeyboardInterrupt:
    print("\nServer stopped by user.")

finally:
    server.server_close()
    print("Server closed safely.")
```

```python
# Program 12: Ping Sweep
## Statement
Write a Python program to check which hosts in a small authorized
local network range respond to ping requests.
```

```python
import subprocess
import platform

network = "127.0.0."

print("===== LOCALHOST PING SWEEP =====")

for host in range(1, 6):
    ip = network + str(host)

    if platform.system().lower() == "windows":
        command = ["ping", "-n", "1", "-w", "500", ip]
    else:
        command = ["ping", "-c", "1", "-W", "1", ip]

    result = subprocess.run(
        command,
        stdout=subprocess.DEVNULL,
        stderr=subprocess.DEVNULL
    )

    if result.returncode == 0:
        print(ip, "-> ACTIVE")
    else:
        print(ip, "-> No response")
```

```python
# Program 13: Network Sniffing with Scapy
## Statement
Write a Python program using Scapy to capture a small number of packets
from an authorized local interface and display basic packet information.
```

```python
from scapy.all import sniff

packet_count = 5

def show_packet(packet):
    print("\n----- PACKET -----")

    if packet.haslayer("IP"):
        print("Source      :", packet["IP"].src)
        print("Destination :", packet["IP"].dst)
        print("Protocol    :", packet["IP"].proto)
    else:
        print("Non-IP packet detected.")

print("Starting safe packet capture...")
print("Capturing", packet_count, "packets only.")

sniff(count=packet_count, prn=show_packet)

print("\nCapture completed.")
```

```python
# Program 16: Socket Server Framework
## Statement
Write a reusable TCP socket server framework that accepts client
connections and processes simple commands.
```

```python
import socket

HOST = "127.0.0.1"
PORT = 9000

def process_message(message):
    if message.lower() == "status":
        return "Server Status: ONLINE"

    if message.lower() == "help":
        return "Available commands: status, help, exit"

    if message.lower() == "exit":
        return "Connection closing..."

    return "Unknown command."


server = socket.socket(socket.AF_INET, socket.SOCK_STREAM)
server.bind((HOST, PORT))
server.listen(1)

print(f"Socket framework running on {HOST}:{PORT}")

connection, address = server.accept()

print("Client connected:", address)

while True:
    data = connection.recv(1024)

    if not data:
        break

    message = data.decode().strip()
    response = process_message(message)

    connection.send(response.encode())

    if message.lower() == "exit":
        break

connection.close()
server.close()

print("Server stopped.")
```

```python
# Program 17: HTTP Client with Sockets
## Statement
Write a Python program that creates a raw TCP socket and sends
a basic HTTP GET request to a local HTTP server.
```

```python
import socket

HOST = "127.0.0.1"
PORT = 8080

request = (
    "GET / HTTP/1.1\r\n"
    f"Host: {HOST}\r\n"
    "Connection: close\r\n"
    "\r\n"
)

client = socket.socket(socket.AF_INET, socket.SOCK_STREAM)
client.settimeout(5)

try:
    client.connect((HOST, PORT))
    client.sendall(request.encode())

    response = b""

    while True:
        data = client.recv(4096)

        if not data:
            break

        response += data

    print("===== HTTP RESPONSE =====")
    print(response.decode(errors="ignore"))

except ConnectionRefusedError:
    print("Start the local HTTP server first.")

finally:
    client.close()
```

```python
# Program 18: Advanced Exception Handling
## Statement
Write a Python program demonstrating multiple exception types,
custom exceptions, and the finally block.
```

```python
class WeakPasswordError(Exception):
    """Raised when a password does not meet the required length."""
    pass


def validate_password(password):

    if len(password) < 8:
        raise WeakPasswordError(
            "Password must contain at least 8 characters."
        )

    if not any(char.isdigit() for char in password):
        raise ValueError(
            "Password must contain at least one number."
        )

    return True


try:
    password = input("Enter password: ")

    validate_password(password)

    print("Password passed basic validation.")

except WeakPasswordError as error:
    print("Security Error:", error)

except ValueError as error:
    print("Validation Error:", error)

except KeyboardInterrupt:
    print("\nInput cancelled by user.")

finally:
    print("Validation process completed.")
```

```python
class SecurityScanner:

    def __init__(self, target):
        self.target = target
        self.findings = []

    def add_finding(self, finding):
        self.findings.append(finding)

    def report(self):
        print("===== SECURITY REPORT =====")
        print("Target:", self.target)

        if not self.findings:
            print("No findings recorded.")
            return

        for number, finding in enumerate(self.findings, start=1):
            print(f"{number}. {finding}")
```
